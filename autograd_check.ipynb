```python
import torch
torch.manual_seed(47)
```

Scalar input: $f(x)=x^3sin(x)+e^{x/2}$

```python
x = torch.tensor(1.2, dtype=torch.float64, requires_grad=True)

def f(x):
    return x**3 * torch.sin(x) + torch.exp(x / 2)

# TODO: derive df/dx by hand
hand_gradient = 5.56362
print(hand_gradient)

# TODO: compute with autograd
result = f(x)
result.backward()
autograd_gradient = x.grad
print(autograd_gradient)

# TODO: compute central finite difference
h = 1e-5
finite_difference = (f(x + h) - f(x))/h
print(finite_difference)
```

Vector input

Let $v=(x,y,z)$:
$$
g(x,y,z)=x^2y+\sin(yz)+e^{x-z}+\frac{1}{2}z^2
$$

```python
v = torch.tensor(
    [0.7, -1.2, 0.5],
    dtype=torch.float64,
    requires_grad=True,
)

def g(v):
    x, y, z = v
    return (
        x**2 * y
        + torch.sin(y * z)
        + torch.exp(x - z)
        + 0.5 * z**2
    )

# TODO: derive all three partial derivatives by hand
hand_gradient = torch.tensor([
    -0.45859724184,
    0.902667807455,
    -1.71180549605,
], dtype=torch.float64)
print(hand_gradient)

# TODO: compute with autograd
result = g(v)
result.backward()
autograd_gradient = v.grad
print(autograd_gradient)

# TODO: perturb one vector component at a time
with torch.no_grad():
    h = 1e-5
    finite_difference_gradient = torch.tensor([
        (g((v[0] + h, v[1], v[2])) - g(v))/h,
        (g((v[0], v[1] + h, v[2])) - g(v))/h,
        (g((v[0], v[1], v[2] + h)) - g(v))/h
    ])
    print(finite_difference_gradient)
```
