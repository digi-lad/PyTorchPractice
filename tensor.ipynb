```python
import torch
torch.manual_seed(47)
```

# Creating tensors & Tensors' Data Types

```python
x = torch.zeros(2, 3)
y = torch.ones(2, 3)
z = torch.ones(2, 3, dtype = torch.int)

print(x)
print(y)
print(z)
print(x.size())
print(y.dtype)
print(z.dtype)
```

```python
import numpy as np

data = np.array([1, 2, 3])

t1 = torch.Tensor(data) # Constructor
t2 = torch.tensor(data) # Factories
t3 = torch.as_tensor(data) # Factories
t4 = torch.from_numpy(data) # Factories

# Factories ALLOW to set target dtype, Constructor does NOT
```

```python
print(t1, t1.dtype)
print(t2, t2.dtype)
print(t3, t3.dtype)
print(t4, t4.dtype)

print(torch.get_default_dtype())
```

```python
# Try modify numpy array
data[0] = 0
data[1] = 0
data[2] = 0

print(t1, t1.dtype) # Copy
print(t2, t2.dtype) # Copy
print(t3, t3.dtype) # Share memory
print(t4, t4.dtype) # Share memory

# GOTO: torch.tensor() OR torch.as_tensor()
```

# Basic Operations

```python
a = torch.rand(2, 2)
b = torch.rand(2, 2)

print(a)
print(b)

c = a + b
print(c)

c = torch.add(a, b)
print(c)

b.add_(a)
print(b)

# Similar for subtraction, multiplication, division
```

# Slicing & Indexing

```python
t = torch.rand(5, 3)
print(t)
print(t[:, 0])
print(t[1, :])
print(t[2, 2]) # return TENSOR, not VALUE
print(t[2, 2].item()) # Can only be use if there is ONE element in tensor
```

# Reshaping

```python
q = torch.rand(4, 4)
print(q)

p = q.view(16) # 16 = 4*4
print(p)

p = q.view(-1, 8) # -1 -> PyTorch automatically finds suitable dimension: 2
print(p)
print(p.size())

p = q.reshape(8, 2)
print(p)
```

# Numpy Conversion

```python
import numpy as np
```

```python
m = torch.ones(5)
print(m)

n = m.numpy()
print(n)

m.add_(1)
print(m)
print(n)

# On CPU, change one -> the other will be changed (same memory)
# So: works if:
# 1. numpy -> tensor -> tensor.cuda()
# 2. tensor (currently cuda()) -> tensor (cpu) -> numpy
```

# GPU Acceleration

```python
if torch.cuda.is_available():
    device = torch.device("cuda")
    
    # 1
    k = torch.ones(5, device=device)

    # 2
    h = torch.ones(5)
    h = h.to(device)

    g = k + h
    print(g)
```
