```python
import torch
torch.manual_seed(47)
```

```python
t1 = torch.tensor(1)
# dtype: int64; shape: []; device: cpu
print(t1.dtype, t1.shape, t1.device)

t2 = torch.tensor([1.0, 2.0, 3.0])
# dtype: float32; shape: [3]; device: cpu
print(t2.dtype, t2.shape, t2.device)

t3 = torch.tensor([-1, -2, -3])
# dtype: int64; shape: [3]; device: cpu
print(t3.dtype, t3.shape, t3.device)

t4 = torch.tensor([[1, 2, 3], [4, 5, 6]])
# dtype: int64; shape: [2, 3]; device: cpu
print(t4.dtype, t4.shape, t4.device)

t5 = t2 * t3
# dtype: float32; shape: [3]; device: cpu
print(t5.dtype, t5.shape, t5.device)

t5 = t5.cuda()
# device: cuda:0
print(t5.device)
```
